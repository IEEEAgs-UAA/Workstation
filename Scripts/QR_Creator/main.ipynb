```python
import random
import string
import pandas as pd
import os
import qrcode

# ==============================================================================
# CONFIGURACIÓN DEL SISTEMA DE VOTACIÓN IEEE UAA
# ==============================================================================
# 1. Remplaza esta URL con tu enlace prellenado real obtenido de Google Forms.
# Debe terminar justo en el signo '=' de la entrada de tu campo 'Ingresa QR'.
# Ejemplo: https://docs.google.com/forms/d/e/1FAIpQLSc.../viewform?usp=pp_url&entry.1234567=
URL_BASE_FORMULARIO = "https://docs.google.com/forms/d/e/1FAIpQLSdy92iaHQ51Skjk4oGmAuowC6GudSMtzKnv4qxZz16k5_dmaA/viewform?usp=pp_url&entry.2072139919="

CANTIDAD_TOKENS = 100
CARPETA_QRS = "qrs_votacion"
ARCHIVO_EXCEL = "lista_maestra_tokens.xlsx"

# Crear carpetas si no existen
os.makedirs(CARPETA_QRS, exist_ok=True)

print("[-] Generando tokens únicos...")
tokens = set()
while len(tokens) < CANTIDAD_TOKENS:
    # Genera códigos alfanuméricos de 8 caracteres altamente seguros
    token = ''.join(random.choices(string.ascii_uppercase + string.digits, k=8))
    tokens.add(token)

lista_tokens = list(tokens)

print("[-] Construyendo enlaces personalizados y códigos QR...")
datos_excel = []

for i, token in enumerate(lista_tokens, start=1):
    # Crear el enlace único para este token específico
    enlace_prellenado = f"{URL_BASE_FORMULARIO}{token}"

    # Guardar en la lista para el archivo maestro de Excel
    datos_excel.append({
        "ID": i,
        "Token": token,
        "Enlace Prellenado": enlace_prellenado,
        "Estado": "FALSO" # FALSO significa que aún no ha sido usado en el evento
    })

    # Configurar y generar el código QR con alta calidad
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(enlace_prellenado)
    qr.make(fit=True)

    # Crear la imagen del código QR (Blanco y Negro formal)
    img_qr = qr.make_image(fill_color="black", back_color="white")

    # Guardar la imagen con el nombre del token correspondiente
    nombre_archivo_qr = os.path.join(CARPETA_QRS, f"QR_Boleto_{i:03d}_{token}.png")
    img_qr.save(nombre_archivo_qr)

# Guardar los datos en el Excel maestro
df = pd.DataFrame(datos_excel)
df.to_excel(ARCHIVO_EXCEL, index=False)

print(f"\n[¡ÉXITO!] Proceso finalizado:")
print(f" 1. Se ha creado el archivo Excel maestro: '{ARCHIVO_EXCEL}'")
print(f" 2. Se han guardado {CANTIDAD_TOKENS} imágenes QR individuales en la carpeta: '{CARPETA_QRS}/'")
print("\nRecuerda modificar la variable 'URL_BASE_FORMULARIO' con tus datos reales antes de ejecutar.")
```
